```python
1. Create a DataFrame named employees with the following data:
data = { 'EmployeeID': ['E01', 'E02', 'E03', 'E04'], 'Name': ['John Doe', 'Jane Smith', 'Emily Jones', 'Chris Johnson'], 'Department': ['HR', 'Finance', 'IT', 'Marketing'] }
Set the EmployeeID column as the index of the DataFrame.
Display only the employee with ID 'E03'.
Reset the index without removing it from the DataFrame and display the result.
```

```python
import pandas as pd

# create DataFrame
data = {
    'EmployeeID': ['E01', 'E02', 'E03', 'E04'],
    'Name': ['John Doe', 'Jane Smith', 'Emily Jones', 'Chris Johnson'],
    'Department': ['HR', 'Finance', 'IT', 'Marketing']
}

employees = pd.DataFrame(data)

# set EmployeeID as index
employees.set_index('EmployeeID', inplace=True)

# display employee with ID 'E03'
employee_e03 = employees.loc['E03']
print(employee_e03)

# reset index without dropping it
employees_reset = employees.reset_index(drop=False)
print(employees_reset)
```

```python
2. Create a DataFrame named sales_data containing product sales data with missing values:
sales_data = pd.DataFrame({ 'Month': ['January', 'February', 'March', 'April'], 'Sales': [200, None, 250, None] })
Calculate the average sales value using the available data.
Fill the missing values in the Sales column with the calculated average.
```

```python
import pandas as pd

# create DataFrame
sales_data = pd.DataFrame({
    'Month': ['January', 'February', 'March', 'April'],
    'Sales': [200, None, 250, None]
})

# calculate mean sales (ignores None / NaN by default)
mean_sales = sales_data['Sales'].mean()

# fill missing values with the mean
sales_data['Sales'] = sales_data['Sales'].fillna(mean_sales)

print(sales_data)
```

```python
4. Based on the Netflix data, create a new column in which you assign the label 'to be archived' to titles with a release year earlier than 2015.
Otherwise, assign the label 'current title'.
```

```python
import pandas as pd

netflix_data['status'] = netflix_data['release_year'].apply(
    lambda x: 'to be archived' if x < 2015 else 'current title'
)

print(netflix_data)
```

```python
6. Using the Library_data file, create a DataFrame and perform the following tasks:

- Group the data by the genre column and display the average book price for each genre

- Filter the data to include only records where the book condition is 'good' and create a new DataFrame

- Calculate the average, minimum, maximum, and median book price for the entire dataset

- Interpret the results in a few sentences
```

```python
import pandas as pd

# load data
df = pd.read_csv('sample_data/Library_Data - Library_Data.csv')

# normalize column names
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(' ', '_')
)

# check columns
print(df.columns)

# 1. average book price by genre
avg_price_by_genre = df.groupby('genre')['price'].mean()
print("\nAverage price by genre:")
print(avg_price_by_genre)

# 2. filter books with condition 'good'
good_books_df = df[df['book_condition'] == 'Good']
print("\nBooks in good condition:")
print(good_books_df.head())

# 3. overall price statistics
mean_price = df['price'].mean()
min_price = df['price'].min()
max_price = df['price'].max()
median_price = df['price'].median()

print("\nOverall price statistics:")
print(f"Mean price: {mean_price}")
print(f"Minimum price: {min_price}")
print(f"Maximum price: {max_price}")
print(f"Median price: {median_price}")
```
