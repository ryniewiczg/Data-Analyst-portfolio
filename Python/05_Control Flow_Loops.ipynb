FOR LOOP





```python
#Example 1 – Iterating over a list of strings
sports_list = ["swimming", "tennis", "football", "basketball"]

for sport in sports_list:
    print(f"My favorite sport is {sport}")
```

```python
#Example 2 – Iterating over a list of numbers

numbers_list = [1, 2, 3, 4, 5, 6]

for number in numbers_list:
    print(number + 1)
```

```python
#Example 3 – Using range()

for element in range(5):
    print(f"Number is: {element}")
```

```python
#Example 4 – Looping through dictionary keys

user_information = {"name": "John", "surname": "Smith", "age": "40"}

for key in user_information.keys():
    print(key)
```

```python
#Example 5 – Looping through dictionary items

user_information = {"name": "John", "surname": "Smith", "age": "40"}

for item in user_information.items():
    print(item)
```

```python
#Example 6 – Looping through dictionary values
user_information = {"name":"John","surname":"Smith","age":"40"}

for value in user_information.values():
    print(value)
```

```python
#Example 7 – Summing elements in a list
list_numbers = [20, 40, 60, 80]
result = 0

for element in list_numbers:
    result = result + element
    print(result)

```

```python
#Example 8 – Length of strings
fruits = ["bananas", "apples", "oranges"]

for fruit in fruits:
    print(f"The word {fruit} has length {len(fruit)}")
```

```python
#Example 9 – Counting elements
elements_list = [10, 20, 30, "A", "B", "C"]
count = 0

for element in elements_list:
    count = count + 1
    print(count)
```

```python
#Example 10 – Basic while loop
number = 1

while number <= 5:
    print(number)
    number = number + 1
```

```python
#Example 11 – Doubling value with a condition
count = 1
max_iterations = 100

while count < 100:
    if count > max_iterations:
        print("Iteration limit reached")
        break
    count = count * 2
    print(count)
```

```python
#Example 12 – Using break
lista = range(10)

for number in range(10):
    if number == 5:
        break
    print(number)
```

```python
#Example 13 – Detecting negative numbers

numbers = [1, 2, 3, 4, 5, -6, 7, 8]

for number in numbers:
    if number > 0:
        print(number)
    elif number < 0:
        print("Negative number detected:", number)
        break
```

```python
#Example 14 – Using continue
for number in range(10):
    if number % 2 == 0:
        continue
    print(number)
```
