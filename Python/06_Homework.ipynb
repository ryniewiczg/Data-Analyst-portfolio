```python
Write a program that asks the user for the number of points they scored (range 0-100).
Based on the number of points, classify the user:

0-49 points:
"Unfortunately, you did not qualify."

50-69 points:
"Congratulations, you qualified for the basic level!"

70-84 points:
"Great job, you qualified for the intermediate level!"

85-100 points:
"Excellent, you qualified for the advanced level!"
```

```python
points = int(input("Enter the number of points (0-100): "))

if 0 <= points <= 49:
    print("Unfortunately, you did not qualify.")
elif 50 <= points <= 69:
    print("Congratulations, you qualified for the basic level!")
elif 70 <= points <= 84:
    print("Great job, you qualified for the intermediate level!")
elif 85 <= points <= 100:
    print("Excellent, you qualified for the advanced level!")
else:
    print("Error: please enter a number between 0 and 100.")
```

```python
Ask the user to enter two numbers.
Try to divide the first number by the second one.
Use a try / except block to catch the division by zero error.
Print the result of the division or an error message.
```

```python
try:
    a = float(input("Enter the first number: "))
    b = float(input("Enter the second number: "))

    result = a / b
    print(f"Division result: {result}")

except ZeroDivisionError:
    print("Error: division by zero is not allowed.")
```

```python
You are given a list of integers:
[1, 3, 4, 5, 64, 78, 6, 55, 66, 75, 31, 22]
Use a for loop to find and print only the even numbers from the list.
```

```python
liczby = [1, 3, 4, 5, 64, 78, 6, 55, 66, 75, 31, 22]

for liczba in liczby:
    if liczba % 2 == 0:
        print(liczba)
```

```python
Write a function double_number that takes a number as an argument and returns its doubled value.
Call the function for several numbers and print the results.
```

```python
def double_number(number):
    return number * 2


print(double_number(2))
print(double_number(5))
print(double_number(10))
print(double_number(-3))
```

```python
Write a function is_even that takes a number as an argument and returns True if the number is even, and False otherwise.
Call the function for several example numbers and print the results.
```

```python
def is_even(number):
    return number % 2 == 0


print(is_even(2))   # True
print(is_even(7))   # False
print(is_even(0))   # True
print(is_even(-4))  # True
```
