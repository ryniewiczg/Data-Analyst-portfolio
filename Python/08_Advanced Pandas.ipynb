```python
kategorie = pd.DataFrame({
    'Name': ['Alice', 'Bob', None, 'David'],
    'Category': ['A', 'B', 'C', 'A']
})
kategorie
```

```python
# mapping categories


mapowanie={
    'A':'Premium',
    'B':'Medium',
    "C":'Economy'
}
kategorie['Upper_Level'] = kategorie['Category'].map(mapowanie)
kategorie
```

```python
# mapping using a function


def funkcja_mapujaca(kategorie):
  if kategorie['Category'] == 'A':
    return "Kategoria A"
  else:
    return "Kategoria B lub C"

kategorie['Highest_Level'] = kategorie.apply(funkcja_mapujaca, axis=1)
kategorie
```
