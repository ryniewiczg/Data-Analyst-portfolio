Exercise 1: Create variables age, height, and name that store respectively: age (integer), height (floating-point number), and name (string).
Print a sentence containing all these variables, e.g.:
"John is 30 years old and 1.75 m tall."

```python
age = 30        # integer
height = 1.75  # float
name = "John"  # string

print(f"{name} is {age} years old and {height} m tall.")
```

Exercise 5: Write a program that asks the user to enter their year of birth.
Calculate and print how old the user will be in the current year.
Use the current year as a constant, e.g. current_year = 2026.

Print the result in the following format:
"In 2026, you will be X years old."

```python
birth_year = input("Enter your year of birth: ")
current_year = 2026

birth_year = int(birth_year)
difference = current_year - birth_year

print(f"In 2026, you will be {difference} years old.")
```
