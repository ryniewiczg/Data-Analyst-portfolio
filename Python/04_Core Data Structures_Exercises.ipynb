Exercise 1: Define a variable celsius with an example temperature value in degrees Celsius.
Calculate the equivalent in degrees Fahrenheit using the formula F = C × 1.8 + 32.
Print the result in the following format:
"[celsius] degrees Celsius is [fahrenheit] degrees Fahrenheit."

```python
celsius = 32.5
fahrenheit = celsius * 1.8 + 32

print(f"{celsius} degrees Celsius is {fahrenheit} degrees Fahrenheit.")
```

Exercise 2:
Define three variables with grades: grade1, grade2, grade3.
Calculate the average of these grades.
Print the result in the following format:
"The average of grades [grade1], [grade2], [grade3] is [average]."

```python
grade1 = 4
grade2 = 3
grade3 = 6

average = round((grade1 + grade2 + grade3) / 3, 2)

print(f"The average of grades {grade1}, {grade2}, {grade3} is {average}.")

```

Exercise 3: Create a list containing different data types, add new elements to it, modify at least one of the existing elements, and then remove any element.
Finally, print the final version of the list and explain what was changed.

```python
random_list = [1, 2, 0, "car", 1.23]

random_list.append(2)  # add 2 to the end of the list
print(random_list)

random_list[2] = "change"  # modify an element using its index
print(random_list)

random_list.remove("car")  # remove an element by value
print(random_list)
```

Exercise 4: Create a list containing numbers and words, e.g. [3, 1, 4, "apple", "orange"].
Sort the list of numbers and print the result.
Note: remember to remove non-numeric elements before sorting.

```python
new_list = [5, 4, 3, 2, 1, "house", "room"]

del new_list[5:7]          # remove non-numeric elements
new_list.sort(reverse=True)  # sort the numbers in descending order

print(new_list)
```

Exercise 5: Create a list of numbers from 1 to 20.
Use slicing to extract numbers from 5 to 15.
Then extract every third element from this sublist.
Finally, reverse the entire list using slicing ([::-1]).

```python
numbers = [i for i in range(1, 21)]

sublist = numbers[4:15]        # numbers from 5 to 15
every_third = sublist[::3]    # every third element
reversed_list = numbers[::-1] # reverse the list

print("Sublist 5–15:", sublist)
print("Every third element:", every_third)
print("Reversed list:", reversed_list)
```
