```python
# reversing the order of elements in a list

number_list = [1, 2, 3, 4, 6]
number_list.reverse()
print(number_list)
```

```python
# sorting a list in descending order; if we omit the 'reverse' argument, the list will be sorted in ascending order

number_list = [5, 2, 8, 1, 9, 6]
number_list.sort(reverse=True)
print(number_list)
```

```python
# adding a single element to a list

number_list = [5, 2, 8, 1, 9, 6]
number_list.append(10)
print(number_list)
```

```python
# adding multiple elements to a list

number_list = [5, 2, 8, 1, 9, 6]
number_list.extend([1, 2, 3, 5])
print(number_list)


```

```python
# adding an element at a specific index

number_list = [5, 2, 8, 1, 9, 6]
number_list.insert(0, 11)
print(number_list)

```

```python
# assigning a value to a list element by index

number_list = [5, 2, 8, 1, 9, 6]
number_list[0] = "number"
print(number_list)
```

```python
# removing a list element by index

number_list = [5, 2, 8, 1, 9, 6]
del number_list[1]
print(number_list)
```

```python
# removing a list element by value

number_list = [5, 2, 8, 1, 9, 6]
number_list.remove(2)
print(number_list)
```

```python
# removing an element and storing the removed value

number_list = [5, 2, 8, 1, 9, 6]
x = number_list.pop(0)
print(number_list)
print("Removed value:", x)

```

```python
# list slicing (extracting elements)

number_list = [5, 2, 8, 1, 9, 6, 4, 5, 9, 11, 12, 13]

# from the beginning to index 4 (elements 0–4)
print(number_list[:5])

# from index 3 to the end
print(number_list[3:])

# between indices 2 and 5 (elements 2–4)
print(number_list[2:5])

# the entire list (copy of the list)
print(number_list[:])




```

```python
# finding the index of a list element by its value

number_list = [5, 2, 8, 1, 9, 6, 4, 5, 9, 11, 12, 13]
print(number_list.index(5))
```

```python
# adding multiple elements to a dictionary (values are added using update())

dictionary = {"name": "John", "age": 30}
dictionary.update({"city": "Warsaw", "postal_code": "01-001"})
print(dictionary)
```

```python
# displaying and retrieving all dictionary keys

dictionary = {"name": "John", "age": 30, "city": "Warsaw"}
keys = dictionary.keys()
print(keys)
```

```python
# displaying and retrieving all dictionary values

dictionary = {"name": "John", "age": 30, "city": "Warsaw"}
dictionary2 = {"name": "Adam", "age": 31, "city": "Gdansk"}

values = dictionary.values()
values2 = dictionary2.values()

print(values, values2)
```

```python
# displaying and retrieving all key–value pairs from a dictionary

dictionary = {
    "name": "Jan",
    "age": 30,
    "city": "Warsaw",
    "name": "Adam",
    "age": 31,
    "city": "Gdansk"
}

pairs = dictionary.items()
print(pairs)

```

```python
# packing (creating a tuple from multiple elements)

tuple_example = 1, "apple", 1.12
print(tuple_example)

```

```python
# swapping values (changing the order)

x = "apple"
y = "banana"

x, y = y, x
print(x, y)
```

```python
# unpacking (assigning tuple elements to variables)

tuple_example = 1, "apple", 1.12
a, b, c = tuple_example

print(a)
print(b)
print(c)

```

```python
# counting how many times an element appears in a tuple

tuple_test = (1, 2, 3, 4, 4, 3, 2, 2, 2)
count = tuple_test.count(2)
print(count)
```

```python
# selecting a tuple element by index

tuple_test = (1, 2, 3, 4, 4, 3, 2, 2, 2)
print(tuple_test[2])

```

```python
# selecting tuple elements by index range (slicing)

tuple_test = (1, 2, 3, 4, 4, 3, 2, 2, 2)
print(tuple_test[1:3])
```

```python
# finding the index of an element in a tuple by its value

tuple_test = (1, 2, 3, 4, 4, 3, 2, 2, 2)
tuple_index = tuple_test.index(4)
print(tuple_index)
```

```python
# adding an element to a set

set_example = {1, 2, 3}
set_example.add(4)
print(set_example)
```

```python
# removing a value from a set

set_example = {1, 2, 3}
set_example.remove(2)
print(set_example)
```

```python
# merging sets (union)

a = {1, 2, 3}
b = {1, 2, 3, 4, 5, 6}

c = a.union(b)
print(c)
```

```python
# common elements of sets (intersection)

a = {1, 2, 3}
b = {1, 2, 3, 4, 5, 6}

c = a.intersection(b)
print(c)
```

```python
# set difference (subtracting sets)

a = {1, 2, 3}
b = {1, 2, 3, 4, 5, 6}

c = b.difference(a)
print(c)
```

```python
# checking whether a set is a subset of another set (True/False)

a = {1, 2, 3}
b = {1, 2, 3, 4, 5, 6}

print(a.issubset(b))
```
