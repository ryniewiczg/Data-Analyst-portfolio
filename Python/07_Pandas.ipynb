```python
#creating a pandas Series from a list

moja_lista=[1,2,4,5,6,7,8,9]
data= pd.Series(moja_lista)
print(data)
```

```python
#creating a Series from a list with indexes

indexed_data = pd.Series([10,20,30,40,50], index=['mon','tue','wed','thu','fri'])
print(indexed_data)
```

```python
#example: indexes help count values regardless of their order

a = pd.Series([1,2], index=["x","y"])
b= pd.Series([6,4], index =["y","x"])
a+b
```

```python
#counting rows by a specific category

df_netflix['type'].value_counts()
```

```python
#nunique – counting unique values in a column

df_netflix['rating'].nunique()
df_netflix['rating'].value_counts()
```

```python
#aggregation – sum

total_sum=df_netflix[['duration_minutes','duration_seasons']].sum()
total_sum
```

```python
#aggregation – mean

table_mean=df_netflix[['duration_minutes','duration_seasons']].mean()
table_mean
```

```python
#aggregation – median; you can also use: min, max, std

table_median=df_netflix[['duration_minutes','duration_seasons']].median()
table_median
```

```python
#aggregation: two columns, multiple calculations

df_netflix.groupby('type')[['duration_minutes','duration_seasons']].agg(['min','max','mean','median'])
```

```python
#correlation matrix #1

correlation= df_netflix[['duration_minutes','duration_seasons']].corr()
correlation
```
